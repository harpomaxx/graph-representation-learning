```python
import pandas as pd
```

```python
captures = ["10","11","12","15","15-2","16","16-2","16-3","17","18","18-2","19","15-3"]
usecols = ["node","ID","OD","IDW","ODW","label"]

cont = 1
for i in captures:
    # features
    features_orig = pd.read_csv(f'/mnt/features-orig/capture201108{i}.binetflow.labels-positive-weights.labeled.csv', sep=",", header=0, usecols=lambda c: c in set(usecols))
    features_prefix = features_orig.copy()
    features_prefix.loc[:,"label"] = features_prefix.loc[:,"label"].replace(to_replace="background", value="normal")
    features_prefix["node"] = str(cont) + "-" + features_prefix["node"].astype(str)
    features_prefix.to_csv(f'/mnt/features-prefix/capture201108{i}_features_prefix.csv', index=False)
    # ncol
    ncol_orig = pd.read_csv(f'/mnt/ncol-orig/capture201108{i}.binetflow.labels-positive-weights.ncol', sep=" ", header=None, names=["source", "target", "weight"])
    ncol_prefix = ncol_orig.copy()
    ncol_prefix["source"] = str(cont) + "-" + ncol_prefix["source"].astype(str)
    ncol_prefix["target"] = str(cont) + "-" + ncol_prefix["target"].astype(str)
    ncol_prefix.to_csv(f'/mnt/ncol-prefix/capture201108{i}_ncol_prefix.ncol', sep=" ", header=False, index=False)
    # incremento contador
    cont += 1
```

```python
# En jorch

captures = ["10","11","12","15","15-2","16","16-2","16-3","17","18","18-2","19","15-3"]
usecols = ["node","ID","OD","IDW","ODW","label"]

cont = 1
for i in captures:
    # features
    features_orig = pd.read_csv(f'capture201108{i}_features.csv', sep=",", header=0, usecols=lambda c: c in set(usecols))
    features_prefix = features_orig.copy()
    features_prefix.loc[:,"label"] = features_prefix.loc[:,"label"].replace(to_replace="background", value="normal")
    features_prefix.loc[:,"label"] = features_prefix.loc[:,"label"].replace(to_replace="botnet", value="infected")
    features_prefix["node"] = str(cont) + "-" + features_prefix["node"].astype(str)
    features_prefix.to_csv(f'features_prefix_BYTES/capture201108{i}_features_prefix.csv', index=False)
    # incremento contador
    cont += 1
```
